## Physics context

Jets are reconstructed proxies for collimated particle flows. In DirectCentauro, clustering is iterative: the C++ algorithm builds pair and beam candidates, selects an action, mutates the event state, and repeats. A selector is therefore one bounded decision component in the loop, not a complete jet algorithm.

For positive candidate distances, the local sampling method assigns probabilities proportional to `d_i^(-2a)`. A larger positive exponent increases preference for smaller distances, but finite sampling is not an exact minimum finder. Exact zero distance takes an explicit bypass.

## Architecture and validation

| Layer | Owns | Public boundary |
| --- | --- | --- |
| C++ DirectCentauro | event state, candidate distances, merge/beam actions, EDM output | sibling EICrecon additions |
| Python local worker | bounded local sampling and one actionable candidate-index proposal plus diagnostics | local socket request/reply |
| Contract | strict schemas and response validation | exact-wire SHA-256 digest binding |

The response digest is SHA-256 of the exact UTF-8 request payload bytes, excluding the JSONL line-feed delimiter. This prevents a validated response from being applied to different request bytes.

**Shadow** records a local selector choice but applies the C++ classical action. **Active** may apply only a schema-valid index whose digest matches the exact request. The bare configuration defaults `quantumFailClosed=false`; `run-active` forces `true`, and `run-shadow` defaults `true` with an exploratory false override. Exact-zero pair distance is applied locally by C++ without a worker request or digest.
